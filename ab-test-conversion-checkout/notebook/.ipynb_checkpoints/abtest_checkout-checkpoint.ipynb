```python
# Так как в предыдущих пунктах мы проверили, что клиент с charge строго имеет событие checkoutview, то
# можно всем клиентам с событием charge присвоить tag контроля или теста (в изначальном датасете эти строки не заполнены)

# Строим маппинг ClientID -> tag по событиям checkoutView
# Если у клиента несколько тегов, берём моду (самый частый)
checkout_tags = (
    df.loc[(df["EventType"] == "checkoutView") & (df["tag"].notna()), ["ClientID", "tag"]]
    .groupby("ClientID")["tag"]
    .agg(lambda x: x.mode().iat[0] if not x.mode().empty else x.iloc[0])
)

# Создаём новый датасет df_final и заполняем tag для пустых случаев по маппингу
df_final = df.copy()
df_final["tag"] = df_final["tag"].fillna(df_final["ClientID"].map(checkout_tags))

# Контрольные проверки
charge_missing_before = df.loc[df["EventType"] == "charge", "tag"].isna().sum()
charge_missing_after = df_final.loc[df_final["EventType"] == "charge", "tag"].isna().sum()
print("charge без tag ДО:", charge_missing_before)
print("charge без tag ПОСЛЕ:", charge_missing_after)
print(df_final["tag"].value_counts(dropna=False))
```

```python
# Расчет CR (уникальные клиенты) по группам

from scipy.stats import norm

# --- Подготовка выборок ---
views = df_final[df_final["EventType"] == "checkoutView"]
charges = df_final[df_final["EventType"] == "charge"]

# Уникальные клиенты по группам
denom = views.groupby("tag")["ClientID"].nunique().rename("unique_viewers")
num = charges.groupby("tag")["ClientID"].nunique().rename("unique_created_donations")

summary = pd.concat([denom, num], axis=1).fillna(0)
summary["unique_viewers"] = summary["unique_viewers"].astype(int)
summary["unique_created_donations"] = summary["unique_created_donations"].astype(int)
summary["CR"] = summary["unique_created_donations"] / summary["unique_viewers"]
```

```python
# --- Доверительные интервалы (Wilson 95%) ---
def wilson_ci(k, n, z=1.96):
    if n == 0:
        return (np.nan, np.nan)
    p = k / n
    denom = 1 + z**2 / n
    center = (p + z**2 / (2 * n)) / denom
    half = z * np.sqrt((p * (1 - p) + z**2 / (4 * n)) / n) / denom
    return (center - half, center + half)

ci_low, ci_high = [], []
for k, n in zip(summary["unique_created_donations"], summary["unique_viewers"]):
    lo, hi = wilson_ci(k, n, z=1.96)
    ci_low.append(lo); ci_high.append(hi)

summary["CR_CI_low"] = ci_low
summary["CR_CI_high"] = ci_high
print(summary)
```

```python
plot_data = summary.reset_index()
x_positions = np.arange(len(plot_data))  # [0, 1]

plt.figure(figsize=(5, 7), dpi=150)

# Рисуем точки и усы
plt.errorbar(
    x=x_positions,
    y=plot_data["CR"],
    yerr=[plot_data["CR"] - plot_data["CR_CI_low"], plot_data["CR_CI_high"] - plot_data["CR"]],
    fmt='o',
    capsize=10,
    elinewidth=2,
    markersize=10,
    color='blue',
    ecolor='black'
)

# Подписи на концах интервалов и в середине
for i, row in plot_data.iterrows():
    # Верхний конец
    plt.text(x_positions[i] + 0.05, row["CR_CI_high"], f"{row['CR_CI_high']:.3f}",
             ha='left', va='bottom', fontsize=9, color='black')
    # Нижний конец
    plt.text(x_positions[i] + 0.05, row["CR_CI_low"], f"{row['CR_CI_low']:.3f}",
             ha='left', va='top', fontsize=9, color='black')
    # Среднее (сама точка)
    plt.text(x_positions[i] + 0.05, row["CR"], f"{row['CR']:.3f}",
             ha='left', va='center', fontsize=9, fontweight='bold', color='blue')

# Настройка осей
plt.xticks(x_positions, plot_data["tag"])
plt.title("CR с 95% доверительными интервалами")
plt.xlabel("Группа")
plt.ylabel("Conversion Rate (CR)")
plt.ylim(0, max(plot_data["CR_CI_high"]) * 1.25)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Теперь посчитаем какого размера должны быть группы, чтобы задетектить MDE в 1 п.п. (с 11.7 до 12.7)
import math

def sample_size_two_proportions(p1, p2, alpha=0.05, power=0.8, ratio=1.0, two_sided=True):
    """
    Минимальные размеры выборок для A/B-теста на разность долей.
    p1: baseline конверсия (контроль), например 0.117
    p2: ожидаемая конверсия (тест), например 0.127
    alpha: уровень значимости (двусторонний, если two_sided=True)
    power: мощность (1 - beta)
    ratio: n_test / n_control (для 50/50 = 1.0)
    two_sided: двухсторонний тест (True) или односторонний (False)
    """
    delta = abs(p2 - p1)
    if delta <= 0:
        raise ValueError("Ожидаемая конверсия p2 должна отличаться от p1.")
    z_alpha = norm.ppf(1 - alpha/2) if two_sided else norm.ppf(1 - alpha)
    z_beta = norm.ppf(power)
    pbar = (p1 + p2) / 2.0
    k = ratio
    term1 = z_alpha * (pbar * (1 - pbar) * (1 + 1/k))**0.5
    term2 = z_beta * (p1 * (1 - p1) + (p2 * (1 - p2)) / k)**0.5
    n_control = ((term1 + term2) ** 2) / (delta ** 2)
    n_test = k * n_control
    return math.ceil(n_control), math.ceil(n_test)

# Пример из задачи:
p1, p2 = 0.117, 0.127   # 11.7% -> 12.7%
alpha, power, ratio = 0.05, 0.80, 1.0
n_control, n_test = sample_size_two_proportions(p1, p2, alpha, power, ratio, two_sided=True)
print({"n_control": n_control, "n_test": n_test, "total": n_control + n_test})
```

```python
# --- Z-тест разницы пропорций test vs control ---
x_control = int(summary.loc["control", "unique_created_donations"])
n_control = int(summary.loc["control", "unique_viewers"])
x_test = int(summary.loc["test", "unique_created_donations"])
n_test = int(summary.loc["test", "unique_viewers"])

p1 = x_test / n_test
p2 = x_control / n_control
p_pool = (x_test + x_control) / (n_test + n_control)
se = np.sqrt(p_pool * (1 - p_pool) * (1 / n_test + 1 / n_control))
z = (p1 - p2) / se
p_value = 2 * (1 - norm.cdf(abs(z)))

print({
    "CR_control": p2,
    "CR_test": p1,
    "z_stat": z,
    "p_value": p_value
})
```

# Из блока 27 можно сделать вывод, что требуется опровергнуть нулевую гипотезу и принять альтернативную, что CR в тестовой группе статистически значимо выше, чем в контрольной

```python
# Формируем данные на уровне пользователя для расчёта ARPU.
# ARPU считаем среди всех, кто видел checkoutView (даже если не донатил).


# 1) Берём всех пользователей, видевших экран чекаута (единица анализа - пользователь)
checkout = (
    df_final[df_final["EventType"] == "checkoutView"][["tag", "ClientID"]]
    .drop_duplicates()
)

# 2) Суммируем все донаты на пользователя внутри группы
donations = df_final[df_final["EventType"] == "charge"]
user_sum = (
    donations.groupby(["tag", "ClientID"])["AmountInUSD"]
    .sum()
    .reset_index()
)

# 3) Левым соединением присоединяем суммы к списку всех «экспонированных» пользователей.
#    Тем, кто не донатил, ставим 0 - это важно для корректного ARPU.
user_revenue = checkout.merge(user_sum, on=["tag", "ClientID"], how="left")
user_revenue["AmountInUSD"] = user_revenue["AmountInUSD"].fillna(0.0)
user_revenue = user_revenue.rename(columns={"AmountInUSD": "RevenuePerUser"})

# Быстрый sanity-check
user_revenue.head()
```

```python
# Выводим 20 самых больших пожертвований
top_20_donations = (
    df_final[df_final["AmountInUSD"].notna()]  # убираем NaN
    .sort_values(by="AmountInUSD", ascending=False)
    .head(20)
)

top_20_donations[["ClientID", "EventTime", "AmountInUSD", "DonationType", "tag"]]
```

Еще раз убеждаемся, что в тестовой группе есть очень большие значения пожертвований, которые сильно отличаются от средней

```python
# ARPU = средний доход на пользователя (среди всех, кто видел checkoutView)

arpu = user_revenue.groupby("tag")["RevenuePerUser"].mean().reset_index()
arpu = arpu.rename(columns={"RevenuePerUser": "ARPU"})
print(arpu)
```

```python
# Копия датасета уровня пользователя
trimmed_groups = []
for g in ["control", "test"]:
    vals = user_revenue.loc[user_revenue["tag"] == g, "RevenuePerUser"]
    
    # Только верхний порог = 99й перцентиль
    high = np.percentile(vals, 99)
    
    # Фильтруем: оставляем всех с RevenuePerUser <= high
    filtered = user_revenue.loc[
        (user_revenue["tag"] == g) &
        (user_revenue["RevenuePerUser"] <= high)
    ].copy()
    
    trimmed_groups.append(filtered)

# Собираем обратно
trimmed_top = pd.concat(trimmed_groups, ignore_index=True)

# Считаем ARPU по группам после отрезания только верхних выбросов
arpu_trimmed_top = trimmed_top.groupby("tag")["RevenuePerUser"].mean()
print("ARPU (обрезано только сверху, выше 99 перцентиля):")
print(arpu_trimmed_top)

# Берём ARPU из расчётов
arpu_control = arpu_trimmed_top["control"]
arpu_test = arpu_trimmed_top["test"]

# Абсолютная разница
diff = arpu_test - arpu_control

# Процентное отличие от контроля
perc_diff = (diff / arpu_control) * 100

print(f"ARPU control: {arpu_control:.2f} $")

print(f"ARPU test: {arpu_test:.2f} $")

print(f"Разница: {diff:.2f} $")

print(f"Процент от контроля: {perc_diff:.2f} %")
```

```python
# --------------------------------------------------------------------
# --------------------------------------------------------------------
# 1) Промежуточные результаты по группам: N пользователей, сумма, ARPU, медиана
# --------------------------------------------------------------------
group_stats = (
    user_revenue.groupby("tag")
    .agg(
        n_users=("ClientID", "nunique"),
        total_revenue=("RevenuePerUser", "sum"),
        ARPU=("RevenuePerUser", "mean"),
        median_RPU=("RevenuePerUser", "median"),
    )
    .reset_index()
)

print("=== Промежуточные результаты по группам (уровень пользователя) ===")
print(group_stats.round({"total_revenue":2, "ARPU":4, "median_RPU":2}))
print()

# Достаём ARPU по группам и разницу test - control
arpu_control = float(group_stats.loc[group_stats["tag"]=="control", "ARPU"])
arpu_test    = float(group_stats.loc[group_stats["tag"]=="test",    "ARPU"])
obs_diff     = arpu_test - arpu_control

print(f"ARPU control = {arpu_control:.4f} $")
print(f"ARPU test    = {arpu_test:.4f} $")
print(f"Наблюдаемая разница (test - control) = {obs_diff:.4f} $")


# --------------------------------------------------------------------
# 2) Перестановочный тест для разницы средних ARPU (test - control)
#    Никаких предположений о нормальности; устойчив к heavy-tail.
# --------------------------------------------------------------------
def permutation_test_ARPU(
    df_users,
    group_col="tag",
    value_col="RevenuePerUser",
    group_a="control",
    group_b="test",
    n_permutations=10000,
    random_state=123,
    return_null=False
):
    rng = np.random.default_rng(random_state)
    vals_a = df_users.loc[df_users[group_col]==group_a, value_col].to_numpy()
    vals_b = df_users.loc[df_users[group_col]==group_b, value_col].to_numpy()

    obs = float(vals_b.mean() - vals_a.mean())

    combined = np.concatenate([vals_a, vals_b])
    n_a = len(vals_a)

    diffs = np.empty(n_permutations, dtype=float)
    for i in range(n_permutations):
        perm = rng.permutation(combined)
        diffs[i] = perm[n_a:].mean() - perm[:n_a].mean()

    # Двусторонний p-value с маленькой поправкой (плюс-один)
    p_value = (np.sum(np.abs(diffs) >= abs(obs)) + 1) / (n_permutations + 1)

    if return_null:
        return obs, p_value, diffs
    return obs, p_value

obs, p_val, diffs = permutation_test_ARPU(
    user_revenue,
    n_permutations=10000,
    random_state=123,
    return_null=True
)

# Границы 95%-критической области нулевого распределения (для наглядности)
crit_low, crit_high = np.percentile(diffs, [2.5, 97.5])

print("=== Перестановочный тест (двусторонний) ===")
print(f"Наблюдаемая разница (test - control) = {obs:.4f} $")
print(f"p-value = {p_val:.4f}")
print(f"95% критическая область нулевого распределения: [{crit_low:.4f}; {crit_high:.4f}]")
print("Если наблюдаемая разница попадает внутрь этого интервала и p-value > 0.05,")
print("значимой разницы не обнаружено.\n")

# --------------------------------------------------------------------
# 3) Визуализация нулевого распределения разницы и отметка наблюдаемой разницы
# --------------------------------------------------------------------
plt.figure(figsize=(7, 4), dpi=130)
plt.hist(diffs, bins=50)
plt.axvline(obs, linestyle="--")      # наблюдаемая разница
plt.axvline(crit_low, linestyle=":")  # границы 95% крит. области
plt.axvline(crit_high, linestyle=":")
plt.title("Permutation null distribution of ΔARPU (test - control)")
plt.xlabel("Разница средних (симулированная при H0)")
plt.ylabel("Частота")
plt.tight_layout()
plt.show()
```

# Мы многократно случайно «перемешиваем» пользователей между test/control, каждый раз считая разницу средних. Это даёт нулевое распределение разницы. Так как наблюдаемая разница лежит внутри типичных значений этого распределения (внутри 95% критической области) и p-value > 0.05, то значит статистически значимой разницы нет.
